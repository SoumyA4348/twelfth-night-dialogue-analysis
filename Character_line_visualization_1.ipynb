```python
import pandas as pd
#required to read CSV's
import matplotlib.pyplot as plt
#required for creating visualizations
```

```python
#Reading our CSV files of cleaned words
speeches =pd.read_csv('data/twelfth_night_speeches.csv')
```

```python
#testing CSV reading
speeches.head()
```

```python
#Calculating and Keeping a Running total of lines per character, using panda's inbuilt summing function

character_line_count=speeches.groupby('speaker')['lines'].sum()
```

```python
#sanity check to make sure lines are being counted correctly
character_line_count
```

```python
#sorting Speakers according to presence
character_line_count.sort_values(ascending=False)
```

```python
#visualizing lines per character using a MatPlotLib bar graph
character_line_count.plot(kind='barh')
#creates a horizontal bar graph
plt.title("Total lines of each character from Shakespeare's 12th Night")
plt.xlabel("Lines spoken")
plt.ylabel("Character name")
plt.show
```

We can then organize this bar graph to scale based on frequency.

```python
character_line_count_sorted=character_line_count.sort_values(ascending=True)
#creates a new variable containing the lines in ascending order of character frequency


character_line_count_sorted.plot(kind='barh')

plt.title("Total lines of each character from Shakespeare's 12th Night,sorted by frequency")
plt.xlabel("Lines spoken")
plt.ylabel("Character name")
plt.show
```

Through this bar graph, we can find several interesting pieces of information:
- The primary speaker in  the play is Toby, the Comic Relief character, followed shortly by Viola, arguably the main character
- The largest gap in line count sits between _____ and ______
- named characters hold the lion's share of spoken lines in the play(The Fool is "Feste" in many vesrions of the text)
- the primary members of the love triangle conflict in the play hold spots #2,3 and #6 respectfully: 1, 4 and 5 are all held by characters involved in the comic relief subplot
